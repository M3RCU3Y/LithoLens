# LithoLens Data Exploration

This notebook gives the team a quick FORCE 2020 inspection path. Keep reusable logic in `src/litholens/`.

```python
from pathlib import Path
import pandas as pd

from litholens.io import load_force_csv
from litholens.schema import available_curve_cols, infer_depth_col, infer_group_col, infer_target_col

DATA_PATH = Path('../data/raw/force_train.csv')
df = load_force_csv(DATA_PATH)
df.shape
```

## FORCE Lithology Distribution

```python
target_col = infer_target_col(df)
well_col = infer_group_col(df)
depth_col = infer_depth_col(df)
curve_cols = available_curve_cols(df)

print(target_col, well_col, depth_col)
df[target_col].value_counts(normalize=True).rename('fraction').to_frame()
```

## Curve Coverage By Well

```python
coverage = df.groupby(well_col)[curve_cols].apply(lambda frame: frame.notna().mean()).reset_index()
coverage.head()
```
